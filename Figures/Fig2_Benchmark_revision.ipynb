```python
# General purpose
import numpy as np
import pandas as pd
from glob import glob

# Metrics
from cellpose_omni.metrics import average_precision

import matplotlib.pyplot as plt
import seaborn as sns

from skimage import io
from MEDUSSA.utils import BorderRemoval
```

### Deconvolved images

```python
basedir = '/Users/reyesmatte/Desktop/MEDUSSA_paper/Fig2/Benchmarking_data/'

manual_masks = sorted(glob(f'{basedir}GT/*.tif'))

### Fine-tuned models

#### CNNs
fmseg = sorted(glob(f'{basedir}Deconvolved/FMseg/*'))
cyto3 = sorted(glob(f'{basedir}Deconvolved/Cyto3/*'))

#### SAM-based
cpsam = sorted(glob(f'{basedir}Deconvolved/CellposeSAM/*'))
microsam = sorted(glob(f'{basedir}Deconvolved/microSAM/*'))

### Off-the shelf, non-ML
morphometrics = sorted(glob(f'{basedir}Deconvolved/Morphometrics/*'))

print(len(manual_masks))

assert len(manual_masks) == len(fmseg), "different amount of deconvolved FM masks segmented with FMSeg than manually annotated masks"
assert len(manual_masks) == len(cyto3), "different amount of deconvolved FM masks segmented with fine-tuned cyto3 than manually annotated masks"

assert len(manual_masks) == len(cpsam), "different amount of deconvolved FM masks segmented with fine-tuned Cellpose-SAM than manually annotated masks"
#assert len(manual_masks) == len(microsam), "different amount of deconvolved FM masks segmented with microSAM than manually annotated masks"

manual_masks = [io.imread(mask) for mask in manual_masks]

fmseg = [BorderRemoval(io.imread(mask)) for mask in fmseg]
cyto3 = [BorderRemoval(io.imread(mask)) for mask in cyto3]

cpsam = [BorderRemoval(io.imread(mask)) for mask in cpsam]
microsam = [BorderRemoval(io.imread(mask)) for mask in microsam]

morphometrics = [BorderRemoval(io.imread(mask)) for mask in morphometrics]
```

```python
thresholds = [0.5, 0.55, 0.6, 0.65, 0.7, 0.75, 0.8, 0.85, 0.9, 0.95, 1]

JI_fmseg, tp, fp, fn = average_precision(manual_masks,fmseg,threshold=thresholds)
f1_fmseg = 2*tp/(2*tp+fn+fp)

JI_cyto3, tp, fp, fn = average_precision(manual_masks,cyto3,threshold=thresholds)
f1_cyto3 = 2*tp/(2*tp+fn+fp)

JI_cpsam, tp, fp, fn = average_precision(manual_masks,cpsam,threshold=thresholds)
f1_cpsam = 2*tp/(2*tp+fn+fp)

JI_microsam, tp, fp, fn = average_precision(manual_masks,microsam,threshold=thresholds)
f1_microsam = 2*tp/(2*tp+fn+fp)

JI_morphometrics, tp, fp, fn = average_precision(manual_masks,morphometrics,threshold=thresholds)
f1_morphometrics = 2*tp/(2*tp+fn+fp)
```

```python
f1s = [f1_fmseg,f1_cpsam,f1_microsam,f1_cyto3,f1_morphometrics]

colors = ['#f0027f','#386cb0','orange','green','gray']

names = ['Omnipose','Cellpose-SAM','microSAM','Cellpose3','Morphometrics']#
labels = ['Omnipose (FMSeg)','Cellpose-SAM','microSAM','Cellpose3','Morphometrics']#

dfs = []

for i in range(len(manual_masks)):

    for f,n in zip(f1s,names):
            
        df = pd.DataFrame()

        df.insert(0,'F1',f[i])
        df.insert(0,'Threshold',thresholds)
        df.insert(0,'Model',n)
        df.insert(3,'Image',i)

        dfs.append(df)

full_df = pd.concat(dfs)
full_df.insert(0,'Image type','Deconvolved')
full_df.to_csv('/Users/reyesmatte/Desktop/MEDUSSA_paper/Raw_data/Fig2_benchmarking_deconvolved.csv')
```

### Plotting

```python
deconvolved_df = pd.read_csv('https://raw.githubusercontent.com/OReyesMatte/MEDUSSA/refs/heads/main/Raw_Data/Fig2_benchmarking_deconvolved.csv')
raw_df = pd.read_csv('https://raw.githubusercontent.com/OReyesMatte/MEDUSSA/refs/heads/main/Raw_Data/Fig2_benchmarking_raw.csv')

full_df = pd.concat([deconvolved_df,raw_df],ignore_index=True)
```

```python
colors = ['#f0027f','#386cb0','orange','green','gray']
colors_raw = ['#f0027f','#386cb0','green','gray']

names = ['Omnipose','Cellpose-SAM','microSAM','Cellpose3','Morphometrics']

thresh = full_df[full_df['Threshold'] == 0.8]

fig,ax = plt.subplots(dpi=300,figsize=(8,12),nrows=2)

sns.lineplot(data=deconvolved_df,x='Threshold',y='F1',hue='Model',palette=colors,hue_order=names,errorbar=None,ax=ax[0],legend=True)
sns.lineplot(data=raw_df,x='Threshold',y='F1',hue='Model',palette=colors,hue_order=names,errorbar=None,ax=ax[0],legend=True,linestyle='--')

ax[0].set_ylim(0,1)
ax[0].set_ylabel('F1 score',fontname='Arial',fontsize=15)
ax[0].set_yticklabels(labels=ax[0].get_yticklabels(),fontsize=15)
ax[0].set_xlim(0.5,1)
ax[0].set_xlabel('IoU threshold',fontname='Arial',fontsize=15)
ax[0].set_xticklabels(labels=ax[0].get_xticklabels(),fontsize=15)
ax[0].legend(labels=names,title='Model')

for i,m in enumerate(names):

    subdf = thresh[thresh['Model'] == m]
    sns.violinplot(data=subdf,x=i,y='F1',inner='quart',hue='Image type',cut=0,hue_order=['Raw','Deconvolved'],split=True,gap=0.1,palette=[colors[i],colors[i]],legend=False,ax=ax[1])

ax[1].set_xticklabels(names)
#ax[1].set_ylim(0,1)
ax[1].set_ylabel('F1 score',fontname='Arial',fontsize=15)
ax[1].set_yticklabels(labels=ax[1].get_yticklabels(),fontsize=15)
ax[1].set_xlabel('')
ax[1].set_xticklabels(labels=ax[1].get_xticklabels(),fontsize=15)

alphas = [0.6,1,0.6,1,0.6,1,0.6,1,0.6,1]
linestyles = ['--','-','--','-','--','-','--','-','--','-']

for i,violin in enumerate(ax[1].collections):
    violin.set_alpha(alphas[i])
    violin.set_linestyle(linestyles[i])
    violin.set_linewidth(1.5)

    if i%2 == 0:
        violin.set_edgecolor('black')
    else:
        violin.set_edgecolor(violin.get_facecolor())

fig.tight_layout()
#fig.savefig('/Users/reyesmatte/Desktop/MEDUSSA_paper/Fig2/benchmarking_all.png')
```
